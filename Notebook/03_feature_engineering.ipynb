```python
#Goal Difference

import pandas as pd
import numpy as np

matches = pd.read_csv("../Data/cleaned/WorldCupMatches_C.csv")


#matches.head()

matches['GoalDifference'] = abs(
    matches['HomeTeamGoals'] - 
    matches['AwayTeamGoals'])

matches[['HomeTeamName','HomeTeamGoals','AwayTeamName','AwayTeamGoals', 'GoalDifference']].head()
```

```python
matches['Winner'] = np.where(matches['HomeTeamGoals']> matches['AwayTeamGoals'],
                             matches['HomeTeamName'], 
                             np.where(
                                 matches['AwayTeamGoals']> matches['HomeTeamGoals'],
                                 matches['AwayTeamName'], 
                                 'Draw'))

matches[['HomeTeamName','HomeTeamGoals','AwayTeamName','AwayTeamGoals', 'Winner']].head()
```

```python
matches['ResultCategory'] = np.where(matches['GoalDifference'] >= 3,
	'Dominant Victory',
    np.where(matches['GoalDifference'] == 0,
	'Close Match',
	'Draw'
))

matches[['HomeTeamName','HomeTeamGoals','AwayTeamName','AwayTeamGoals', 'Winner', 'ResultCategory']].head()
```

```python
matches['HighScoring'] = np.where(matches['GoalDifference'] >= 5, 'Yes', 'No')

matches['HighScoring'].value_counts()
```

```python
KnockOutKeywords = [
    'Final',
    'Semi-finals',
    'Quarter-finals',
    'Round of 16'
]

matches['KnockoutMatch'] = matches['Stage'].apply(lambda x: 'Yes' if any(k in str(x) for k in KnockOutKeywords) else 'No')

matches['KnockoutMatch'].value_counts()
```

```python
matches.to_csv(
    "../Data/cleaned/WorldCupMatches_featured.csv",
    index=False
)
```
