```python
#creating a copy of the dataset for further analysis

cleaned_dataset = matches.copy()

cleaned_dataset.head()
```

```python
cleaned_dataset.shape
```

```python
cleaned_dataset = cleaned_dataset.dropna(how="all")
```

```python
cleaned_dataset.shape
```

```python
cleaned_dataset.info()
```

```python
cleaned_dataset.columns
```

```python
cleaned_dataset.columns = cleaned_dataset.columns.str.strip()

cleaned_dataset.columns = cleaned_dataset.columns.str.replace(" ", "")
cleaned_dataset.columns = cleaned_dataset.columns.str.replace("-", "")
```

```python
cleaned_dataset.columns
```

```python


cleaned_dataset['Year'] = cleaned_dataset['Year'].astype('Int64')
```

```python
cleaned_dataset.info()
```

```python
int_columns = [
    "HomeTeamGoals",
    "AwayTeamGoals"
]

cleaned_dataset[int_columns] = cleaned_dataset[int_columns].astype('Int64')
```

```python
int_columns = [
    "HalftimeHomeGoals",
    "HalftimeAwayGoals"
]

cleaned_dataset[int_columns] = cleaned_dataset[int_columns].astype('Int64')
```

```python
int_columns = [
    "RoundID",
    "MatchID",
    "Attendance"
]

cleaned_dataset[int_columns] = cleaned_dataset[int_columns].astype('Int64')
```

```python
cleaned_dataset.info()
```

```python
cleaned_dataset['HomeTeamName'] = cleaned_dataset['HomeTeamName'].replace({'USA': 'United States'})

cleaned_dataset['AwayTeamName'] = cleaned_dataset['AwayTeamName'].replace({'USA': 'United States'})
```

```python
uniqueCountries = cleaned_dataset['HomeTeamName'].unique()
print(uniqueCountries)
```

```python
cleaned_dataset.head()
```

```python
import pandas as pd

cleaned_dataset['Datetime'] = pd.to_datetime(cleaned_dataset['Datetime'], format='%d %b %Y - %H:%M', errors='coerce')

cleaned_dataset[cleaned_dataset['Datetime'].isnull()]

cleaned_dataset['MatchDate'] = cleaned_dataset['Datetime'].dt.date
cleaned_dataset['MatchTime'] = cleaned_dataset['Datetime'].dt.time
```
